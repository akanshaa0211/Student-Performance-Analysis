```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic Dataset
df = pd.read_csv("titanic.csv")

# Show first 5 rows
print(df.head())

# Dataset shape
print("\nDataset Shape:")
print(df.shape)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing Age values
df['Age'].fillna(df['Age'].median(), inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Survival by Gender
gender_survival = df.groupby('Sex')['Survived'].mean()
print("\nSurvival Rate by Gender:")
print(gender_survival)

# Survival by Passenger Class
class_survival = df.groupby('Pclass')['Survived'].mean()
print("\nSurvival Rate by Class:")
print(class_survival)

# Graph 1
sns.barplot(x='Sex', y='Survived', data=df)
plt.title("Survival Rate by Gender")
plt.show()

# Graph 2
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Passenger Class")
plt.show()

# Graph 3
plt.hist(df['Age'], bins=20, edgecolor='black')
plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

## Introduction

The Titanic dataset contains information about passengers such as age, gender, class, and survival status.

The objective of this project is to analyze survival patterns and understand the factors that influenced survival during the Titanic disaster.
## Data Cleaning

The dataset was checked for missing values and duplicate records.

Missing values were found in the Age, Cabin, and Embarked columns.

The missing Age values were filled using the median age to maintain data quality.
## Data Analysis

The following questions were analyzed:

1. Who survived more: males or females?
2. Did passenger class affect survival chances?
3. What was the age distribution of passengers?

The analysis was performed using Pandas and visualized using Matplotlib and Seaborn.
## Results

### Survival by Gender
Female passengers had a significantly higher survival rate than male passengers.

### Survival by Passenger Class
First-class passengers had the highest survival rate, while third-class passengers had the lowest.

### Passenger Age Distribution
Most passengers were young and middle-aged adults.
## Conclusion

The Titanic survival analysis revealed that gender and passenger class played an important role in survival chances.

Female passengers and first-class passengers were more likely to survive compared to male passengers and third-class passengers.
